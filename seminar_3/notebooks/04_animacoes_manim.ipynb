```python
from pathlib import Path
import os
os.environ['NO_ALBUMENTATIONS_UPDATE'] = '1'
import csv
import json
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'seminar_3/articles').is_dir())
SEMINAR = ROOT / 'seminar_3'
OUT = SEMINAR / 'data/outputs'
PROCESSED = SEMINAR / 'data/processed'
OUT.mkdir(parents=True, exist_ok=True)
PROCESSED.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({'figure.dpi': 110, 'axes.spines.top': False, 'axes.spines.right': False})

def save_csv(path, rows):
    with path.open('w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)
```

```python
import subprocess
import sys

required = ['original.png','mask_original.png','mix_A.png','mix_B.png','mnist_summary.csv']
missing = [name for name in required if not (OUT/name).exists()]
if missing:
    raise FileNotFoundError(f'Execute os notebooks 01–03. Ausentes: {missing}')
subprocess.run([sys.executable, str(SEMINAR/'manim/render.py')], check=True, cwd=ROOT)
```

```python
from IPython.display import Video, Image as DisplayImage, display

MEDIA = OUT / 'manim'
for name in ['TransformacoesRotulos', 'Misturas', 'Resultados']:
    video, gif = MEDIA / f'{name}.mp4', MEDIA / f'{name}.gif'
    assert video.stat().st_size > 1000 and gif.stat().st_size > 1000
    print(name, f'MP4: {video.stat().st_size/1e6:.1f} MB; GIF: {gif.stat().st_size/1e6:.1f} MB')
    display(Video(str(video), embed=True, width=800))
    display(DisplayImage(filename=str(gif), width=600))
```
